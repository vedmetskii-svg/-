## Задание

В файле pr5.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
–10 000 до 10 000 включительно. Определите количество пар, в
которых хотя бы один из двух элементов больше, чем среднее
арифметическое всех чисел в файле, и хотя бы один из двух
элементов оканчивается на 3. В ответе запишите два числа: сначала
количество найденных пар, а затем – максимальную сумму
элементов таких пар. В данной задаче под парой подразумевается
два идущих подряд элемента последовательности.



## Решение

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Можно скопировать и поместить файл с данными любым удобным способом

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

```python
# Создаем пустой список
data = []
# Выполняем операции чтения и заполнение списка
with open('pr5.txt') as F:
  for line in F:
    digit = int(line)
    data.append(digit)

# Посмотрим, что мы считали...
print(len(data))
print(data[:10])
```

Теперь ответим на поставленные вопросы использую список `data`

```python
# Вычислим среднее арифметическое всех чисел в файле
sr = sum(data) / len(data)
print(sr)
```

```python
# Подсчет количества подходящих пар
counter = 0
max_sum = -10000
for i in range(len(data)-1):
  # пара элементов
  x, y = data[i], data[i+1]
  # проверяем, подходит ли пара
  if x>sr and y>sr and str(x+y)[-2:] == '31':
    counter +=1
    max_sum = max(max_sum, x+y)
    # print(x, y, x+y)

print(f'Количество подходящих пар = {counter}')
print(f'Максимальная сумма = {max_sum}')
```

Выполните задание в соответствии с индивидуальным вариантом

```python
# Здесь код индивидуального задания...
sr = sum(data) / len(data)
count = 0
max_sum = -10000
for i in range(len(data)-1):
  a, b = data[i], data[i+1]
  if  a > (sr or b > sr) and (abs(a) %10 == 3 or abs(b) %10 == 3):
    count += 1
    max_sum = max(max_sum, a+b)
    print(f'Количество подходящих пар = {count}')
print(f'Максимальная сумма = {max_sum}')
```
