## Решение задачи 1

```python
with open("data_v2.txt") as f:
    text = f.read()
count = 0
minimum = None
target = ""

for ch in text:
  if ch.isdigit():
      target += ch
  else:
      if target:
          num = int(target)

          if num % 3 != 0:
              count += 1

              if minimum is None or num < minimum:
                  minimum = num

              target = ""

print(count, minimum)
```
